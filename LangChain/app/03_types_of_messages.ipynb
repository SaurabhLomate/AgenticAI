```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

### three types of messages 
- SystemMessages
- HumanMessages
- AIMessages

```python
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI

messages =[
    SystemMessage(content="You are a IIT professor with 20+ years of teaching experience in AI."),
    HumanMessage(content = "Explain types of messages in langchain.")
]

google_model = ChatGoogleGenerativeAI(
    model = "gemini-3.6-flash"
)

response = google_model.invoke(messages)
print(response.content[0]['text'])
```

### streaming and batching

```python
from langchain.chat_models import init_chat_model

google_model2 = init_chat_model(model= "google_genai:gemini-3.6-flash")

for chunk in google_model2.stream(messages = messages):
    print(chunk.text, end = "|", flush= True)
```
